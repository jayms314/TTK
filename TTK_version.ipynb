## Version Business

```python
from tkinter import *
from tkinter import ttk
import numpy as np

list_coeff1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]

# Création d'une fenêtre
fenetre = Tk()
fenetre.title("Travail sur l'interface graphique - groupe D")
fenetre.geometry("1000x600")

# Création d'un cadre dans la fenêtre
frm = ttk.Frame(fenetre)
frm.grid()
titre = ttk.Label(frm, text = "Saisie des notes")
titre.grid(column = 1, row = 0)


# Liste des matières
matiere = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

# Affichage des matières
for i, mat in enumerate(matiere):
    objet = ttk.Label(frm, text=mat)
    objet.grid(column=0, row=i+1)

casevert = []

for i in range(21):
    objet1 = Entry(frm, bd=10, bg="Green")
    objet1.grid(column=1, row=i+1)
    casevert.append(objet1)

def enregistre():
    somme_normal = 0
    somme_coeff = 0

    for i, case in enumerate(casevert):
        texte = case.get()
        if texte != "":
            note = float(texte)
            coefficient = list_coeff1[i]
            somme_normal = somme_normal + note * coefficient
            somme_coeff = somme_coeff + coefficient

    if somme_coeff != 0:
        moyenne = somme_normal / somme_coeff
        objet2 = ttk.Label(frm, text=f"La moyenne vaut {moyenne}")
        objet2.grid(column=20, row=2)

bouton = Button(frm, bd=5, text="Valider", command=enregistre)
bouton.grid(column=4, row=0, padx =100)

# Boucle principale
fenetre.mainloop()
```

### Version prof

```python
from tkinter import *
from tkinter import ttk
import numpy as np

coeff1 = [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coeff2 = [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coeff3 = [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

fenetre = Tk()
fenetre.geometry("880x800")

frm = ttk.Frame(fenetre)
frm.grid()
titre = ttk.Label(frm, text="Saisie des notes")
titre.grid(column=1, row=0)

matiere = [f"R{i}" for i in range(101,116)] + [f"SAE{i}" for i in range(11,17)]

notes = []

for i, matiere in enumerate(matiere):
    r = ttk.Label(frm, text=matiere)
    r.grid(column=0, row=i+1, padx=10, pady=5)

    saisi = ttk.Entry(frm)
    saisi.grid(column=1, row=i+1, padx=10, pady=5)

    notes.append(saisi)

def cal():
    result = [float(saisi.get()) for saisi in notes]
    moy1 = round(np.average(result, weights=coeff1),3)
    moy2 = round(np.average(result, weights=coeff2),3)
    moy3 = round(np.average(result, weights=coeff3),3)
    results = ttk.Label(frm, text=f"Les moyennes des UE1,UE2,UE3 sont respectivement {moy1}, {moy2}, {moy3}")
    results.grid(column=2, row=1, padx=20)

boutton = ttk.Button(frm, text="valider", command=cal)
boutton.grid(column=2, row=0)

fenetre.mainloop()
```
