## Interface graphique

## Version pour un placement automatique

```python
from tkinter import*

#Création d'une fenétre 
fenetre= Tk()


objet=Label(fenetre, text="Essayons d'ouvrir la fenetre")
objet2=Label(fenetre, text="Ouais tranquille")

#Automatique 
objet.pack()
objet2.pack()

#boucle
fenetre.mainloop()
```

## Version pour un placement controler 

```python
from tkinter import*
from tkinter import ttk

#Création d'une fenétre 
fenetre= Tk()

#Création d'un cadre dans la fenetre
frm=ttk.Frame(fenetre)
frm.grid()

objet=ttk.Label(frm,text="Essayons d'ouvrir la fenetre")

#Affichage
objet.grid(column=0,row=0)


objet1=ttk.Label(frm,text="ouai")
objet1.grid(column=1,row=0)

#boucle
fenetre.mainloop()
```
